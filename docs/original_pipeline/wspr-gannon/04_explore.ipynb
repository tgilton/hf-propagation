```python
# Fetch 2024 space weather from GFZ
import requests

def fetch_gfz_index(index_name: str, year: int) -> pd.DataFrame:
    url = "https://kp.gfz.de/app/json/"
    params = {
        "start":  f"{year}-01-01T00:00:00Z",
        "end":    f"{year}-12-31T23:59:59Z",
        "index":  index_name,
        "status": "def"
    }
    print(f"Fetching {index_name} for {year}...", end=" ")
    response = requests.get(url, params=params, timeout=60)
    response.raise_for_status()
    data = response.json()
    df = pd.DataFrame({
        "time": pd.to_datetime(data["datetime"]),
        index_name: pd.to_numeric(data[index_name], errors="coerce"),
    })
    if "status" in data:
        df["status"] = data["status"]
    df = df.sort_values("time").reset_index(drop=True)
    print(f"{len(df):,} records")
    return df

df_kp  = fetch_gfz_index("Kp",   2024)
df_sfi = fetch_gfz_index("Fobs", 2024)
df_sn  = fetch_gfz_index("SN",   2024)

print(f"\nKp range: {df_kp['Kp'].min():.1f} – {df_kp['Kp'].max():.1f}")
print(f"SFI range: {df_sfi['Fobs'].min():.1f} – {df_sfi['Fobs'].max():.1f}")
```

```python
import matplotlib.patches as mpatches
import matplotlib.lines as mlines

flare_begin = pd.Timestamp("2023-06-20 16:42:00")
flare_peak  = pd.Timestamp("2023-06-20 17:09:00")
flare_end   = pd.Timestamp("2023-06-20 17:26:00")

# Aggregate to 30-minute bins for finer resolution than df_agg
df_june20["time_30m"] = df_june20["time"].dt.floor("30min")
df_june20["hour_f"] = df_june20["time_30m"].dt.hour + df_june20["time_30m"].dt.minute / 60

# Only day paths with midpoint on sunlit hemisphere — SID only affects dayside
df_day = df_june20[
    (df_june20["path_type"] == "day") &
    (df_june20["mid_sza"] < 90) &
    (df_june20["dist_bucket"].isin(["1k-2k", "2k-4k", "4k-8k"]))
].copy()

fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

band_colors = {10: "tomato", 20: "steelblue", 40: "seagreen"}
band_labels = {10: "10m (28 MHz)", 20: "20m (14 MHz)", 40: "40m (7 MHz)"}

for ax, band_m in zip(axes, [10, 20, 40]):
    subset = df_day[df_day["band_m"] == band_m]
    
    if subset.empty:
        ax.text(0.5, 0.5, "No data", transform=ax.transAxes, ha="center")
        continue
    
    # Median SNR per 30-min bin
    hourly = subset.groupby("time_30m")["snr"].agg(["median", "count"])
    hourly = hourly[hourly["count"] >= 10]  # require at least 10 spots
    
    hour_vals = hourly.index.hour + hourly.index.minute / 60
    
    ax.plot(hour_vals, hourly["median"],
            color=band_colors[band_m], linewidth=2, zorder=3)
    
    # Flare window shading
    ax.axvspan(flare_begin.hour + flare_begin.minute/60,
               flare_end.hour + flare_end.minute/60,
               alpha=0.15, color="red", zorder=1, label="Flare window")
    ax.axvline(flare_peak.hour + flare_peak.minute/60,
               color="red", linewidth=1.5, linestyle="--", zorder=2, label="Flare peak")
    
    # Compute pre/post flare baseline
    pre_mask  = (hour_vals >= 14) & (hour_vals < 16.5)
    post_mask = (hour_vals >= 17.5) & (hour_vals < 20)
    flare_mask = (hour_vals >= 16.5) & (hour_vals <= 17.5)
    
    if pre_mask.any():
        pre_snr = hourly["median"][pre_mask.values].mean()
        ax.axhline(pre_snr, color=band_colors[band_m], linewidth=1,
                   linestyle=":", alpha=0.6, label=f"Pre-flare baseline: {pre_snr:.1f} dB")
    
    ax.set_ylabel("Median SNR (dB)")
    ax.set_title(f"{band_labels[band_m]}")
    ax.legend(loc="upper left", fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_ylim(-30, 5)

axes[-1].set_xlabel("Hour UTC")
axes[-1].set_xticks(range(0, 24, 1))
plt.suptitle(f"X1.1 Solar Flare Case Study — June 20, 2023\n"
             f"Flare: 16:42 → 17:09 peak → 17:26 UTC | Day paths 1k-8k km, sunlit midpoint",
             fontsize=12)
plt.tight_layout()
plt.show()
```

```python
from datetime import timedelta

# Ensemble M-class flare analysis
# For each M-class flare on a WSPR day, extract a ±3 hour window
# normalized to pre-flare baseline

def load_day_band(date_str, band_m):
    path = WSPR_DATA_DIR / f"wspr_{date_str}_{band_m}m_geo.parquet"
    if not path.exists():
        return pd.DataFrame()
    df = pd.read_parquet(path)
    df["band_m"] = band_m
    dist_bins   = [0, 500, 1000, 2000, 4000, 8000, 12000]
    dist_labels = ["<500", "500-1k", "1k-2k", "2k-4k", "4k-8k", "8k+"]
    df["dist_bucket"] = pd.cut(df["distance"], bins=dist_bins, labels=dist_labels)
    return df

# M-class flares on WSPR days
df_m = df_mx[
    (df_mx["flare_class"].str[0] == "M") &
    (df_mx["have_wspr"] == True)
].copy()
print(f"M-class flares to analyze: {len(df_m)}")

# For each band, collect normalized flare responses
WINDOW_HOURS = 3
BIN_MINUTES  = 30

results = {10: [], 20: [], 40: []}
band_dist_ensemble = {10: ["1k-2k"], 20: ["1k-2k","2k-4k"], 40: ["<500","500-1k"]}

for _, flare_row in df_m.iterrows():
    peak_time = pd.Timestamp(flare_row["peak"])
    date_str  = peak_time.strftime("%Y-%m-%d")
    
    for band_m in [10, 20, 40]:
        df_day = load_day_band(date_str, band_m)
        if df_day.empty:
            continue
        
        subset = df_day[
            (df_day["path_type"] == "day") &
            (df_day["mid_sza"] < 90) &
            (df_day["dist_bucket"].isin(band_dist_ensemble[band_m]))
        ].copy()
        
        if subset.empty:
            continue
        
        # 30-min bins in ±3 hour window
        subset["time_30m"] = subset["time"].dt.floor("30min")
        window_start = peak_time - pd.Timedelta(hours=WINDOW_HOURS)
        window_end   = peak_time + pd.Timedelta(hours=WINDOW_HOURS)
        
        subset = subset[
            (subset["time"] >= window_start) &
            (subset["time"] <= window_end)
        ]
        
        if subset.empty:
            continue
        
        hourly = subset.groupby("time_30m")["snr"].agg(["median","count"])
        hourly = hourly[hourly["count"] >= 5]
        
        if len(hourly) < 4:
            continue
        
        # Time offset in minutes from peak
        hourly["offset_min"] = (hourly.index - peak_time).total_seconds() / 60
        
        # Pre-flare baseline: -180 to -30 min before peak
        pre = hourly[(hourly["offset_min"] >= -180) & (hourly["offset_min"] <= -30)]
        if pre.empty:
            continue
        
        baseline = pre["median"].mean()
        hourly["snr_norm"] = hourly["median"] - baseline
        hourly["flare_class"] = flare_row["flare_class"]
        
        results[band_m].append(hourly[["offset_min","snr_norm","count","flare_class"]])

print("\nFlares contributing per band:")
for band_m in [10, 20, 40]:
    print(f"  {band_m}m: {len(results[band_m])} flares")
```

```python
summer_months = [5, 6, 7, 8]
winter_months = [1, 2]

summer_dates = [d for d in fetch_schedule if d.month in summer_months]
winter_dates = [d for d in fetch_schedule if d.month in winter_months]

print(f"Summer days: {len(summer_dates)}")
print(f"Winter days: {len(winter_dates)}")

def load_40m_days(date_list, label):
    frames = []
    for d in date_list:
        date_str = d.strftime("%Y-%m-%d")
        path = WSPR_DATA_DIR / f"wspr_{date_str}_40m_geo.parquet"
        if not path.exists():
            continue
        df = pd.read_parquet(path)
        df["season_label"] = label
        df["date_str"] = date_str
        df["month"] = d.month
        frames.append(df)
    result = pd.concat(frames, ignore_index=True)
    
    # Add distance bucket
    dist_bins   = [0, 500, 1000, 2000, 4000, 8000, 12000]
    dist_labels = ["<500", "500-1k", "1k-2k", "2k-4k", "4k-8k", "8k+"]
    result["dist_bucket"] = pd.cut(result["distance"], bins=dist_bins, labels=dist_labels)
    print(f"{label}: {len(result):,} rows from {len(frames)} days")
    return result

df_summer_40 = load_40m_days(summer_dates, "summer")
df_winter_40 = load_40m_days(winter_dates, "winter")
```

```python
# Fetch 2024 flare catalog for the storm window
import re
from datetime import date as date_type

def parse_events_file(text: str, file_date) -> pd.DataFrame:
    events = []
    for line in text.splitlines():
        if line.startswith('#') or line.startswith(':') or not line.strip():
            continue
        if 'XRA' not in line:
            continue
        try:
            parts = [p for p in line.split() if p != '+']
            xra_idx = parts.index('XRA')
            def hhmm_to_dt(hhmm, d):
                hhmm = hhmm.replace('/', '0')
                try:
                    h, m = int(hhmm[:2]), int(hhmm[2:])
                    return pd.Timestamp(d.year, d.month, d.day, h, m)
                except:
                    return pd.NaT
            particulars   = parts[xra_idx + 2] if xra_idx + 2 < len(parts) else ''
            peak_flux_str = parts[xra_idx + 3] if xra_idx + 3 < len(parts) else ''
            flare_class = particulars if re.match(r'[ABCMX]\d', particulars) else None
            try:
                peak_flux = float(peak_flux_str)
            except:
                peak_flux = None
            events.append({
                "date":        file_date,
                "begin":       hhmm_to_dt(parts[1], file_date),
                "peak":        hhmm_to_dt(parts[2], file_date),
                "end":         hhmm_to_dt(parts[3], file_date),
                "flare_class": flare_class,
                "peak_flux":   peak_flux,
            })
        except:
            continue
    return pd.DataFrame(events)

# Fetch flares for May 7-14 2024
base_url = ("https://www.ngdc.noaa.gov/stp/space-weather/swpc-products/"
            "daily_reports/solar_event_reports")

flare_frames = []
for d in gannon_dates:
    url = f"{base_url}/2024/{d.month:02d}/{d.strftime('%Y%m%d')}events.txt"
    try:
        r = requests.get(url, timeout=30)
        if r.status_code == 200:
            df = parse_events_file(r.text, d)
            if not df.empty:
                flare_frames.append(df)
    except:
        pass

df_flares_2024 = pd.concat(flare_frames, ignore_index=True)
df_flares_2024 = df_flares_2024.dropna(subset=["flare_class"])
df_flares_2024 = df_flares_2024.sort_values("peak").reset_index(drop=True)

mx_flares = df_flares_2024[df_flares_2024["flare_class"].str[0].isin(["M","X"])]
print(f"Total flares May 7-14: {len(df_flares_2024)}")
print(f"M+X class flares: {len(mx_flares)}")
print(f"\nX-class flares:")
print(mx_flares[mx_flares["flare_class"].str[0]=="X"][["peak","flare_class","peak_flux"]])
```

```python
# Load all Gannon geo parquets into a single hourly aggregation
# Similar to what we did for 2023 but at hourly resolution across 8 days

dist_bins   = [0, 500, 1000, 2000, 4000, 8000, 12000]
dist_labels = ["<500", "500-1k", "1k-2k", "2k-4k", "4k-8k", "8k+"]

frames = []
for d in gannon_dates:
    date_str = d.strftime("%Y-%m-%d")
    for band_m in BANDS:
        path = WSPR_DATA_DIR / f"wspr_{date_str}_{band_m}m_geo.parquet"
        if not path.exists():
            continue
        df = pd.read_parquet(path)
        df["band_m"] = band_m
        df["dist_bucket"] = pd.cut(df["distance"], bins=dist_bins, labels=dist_labels)
        df["time_1h"] = df["time"].dt.floor("1h")
        
        # Aggregate to hourly
        agg = df.groupby(["time_1h", "dist_bucket", "path_type"], observed=True).agg(
            spot_count    = ("snr", "count"),
            snr_median    = ("snr", "median"),
            path_loss_med = ("path_loss_proxy", "median"),
            power_median  = ("power", "median"),
        ).reset_index()
        agg["band_m"] = band_m
        frames.append(agg)

df_gannon = pd.concat(frames, ignore_index=True)
df_gannon["time_1h"] = pd.to_datetime(df_gannon["time_1h"])

print(f"Gannon aggregated: {len(df_gannon):,} rows")
print(f"Time range: {df_gannon['time_1h'].min()} → {df_gannon['time_1h'].max()}")
print(f"Bands: {sorted(df_gannon['band_m'].unique())}")
```

```python
# Quantify storm impact vs pre-storm baseline
# Pre-storm baseline: May 7-9 (72 hours)
# Storm peak: May 11 00:00 - 12:00 UTC (Kp=9 period)
# Recovery: May 12 12:00 - May 13 23:00 UTC

periods = {
    "Pre-storm\n(May 7-9)":     ("2024-05-07 00:00", "2024-05-10 00:00"),
    "Storm onset\n(May 10)":    ("2024-05-10 17:00", "2024-05-11 00:00"),
    "Storm peak\n(May 11 00-12)": ("2024-05-11 00:00", "2024-05-11 12:00"),
    "Recovery\n(May 12-13)":    ("2024-05-12 12:00", "2024-05-14 00:00"),
}

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

period_colors = ["steelblue", "goldenrod", "tomato", "seagreen"]
x = np.arange(len(periods))
width = 0.25

for col, (metric, ylabel) in enumerate([
    ("spot_count", "Median hourly spot count"),
    ("snr_median", "Median SNR (dB)")
]):
    ax = axes[col]
    
    for i, band_m in enumerate([10, 20, 40]):
        subset = df_gannon[
            (df_gannon["band_m"] == band_m) &
            (df_gannon["dist_bucket"].isin(["1k-2k", "2k-4k"]))
        ]
        
        values = []
        for period_label, (start, end) in periods.items():
            period_data = subset[
                (subset["time_1h"] >= start) &
                (subset["time_1h"] < end)
            ]
            if metric == "spot_count":
                val = period_data.groupby("time_1h")["spot_count"].sum().median()
            else:
                val = period_data["snr_median"].median()
            values.append(val)
        
        offset = (i - 1) * width
        bars = ax.bar(x + offset, values, width,
                      color=band_colors[band_m], alpha=0.8,
                      label=band_labels[band_m])
    
    ax.set_xticks(x)
    ax.set_xticklabels(periods.keys(), fontsize=9)
    ax.set_ylabel(ylabel)
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3, axis="y")

plt.suptitle("Gannon Storm Impact by Phase — DX Paths 1k-4k km",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# Magnetic latitude dependence during the Gannon storm
# Compare high vs low mlat paths during storm peak vs pre-storm

df_gannon_raw = []
for d in gannon_dates:
    date_str = d.strftime("%Y-%m-%d")
    path = WSPR_DATA_DIR / f"wspr_{date_str}_20m_geo.parquet"
    if path.exists():
        df = pd.read_parquet(path)
        df["date_str"] = date_str
        df_gannon_raw.append(df)

df_20m_raw = pd.concat(df_gannon_raw, ignore_index=True)
df_20m_raw["dist_bucket"] = pd.cut(df_20m_raw["distance"],
                                    bins=dist_bins, labels=dist_labels)

# Define periods
pre_mask  = (df_20m_raw["time"] >= "2024-05-07") & \
            (df_20m_raw["time"] < "2024-05-10")
peak_mask = (df_20m_raw["time"] >= "2024-05-11 00:00") & \
            (df_20m_raw["time"] < "2024-05-11 12:00")

# Magnetic latitude bins
mlat_bins   = [20, 40, 55, 90]
mlat_labels = ["Low (<40°)", "Mid (40-55°)", "High (55°+)"]

df_20m_raw["mlat_bin"] = pd.cut(df_20m_raw["mid_mlat"],
                                 bins=mlat_bins, labels=mlat_labels)

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

mlat_colors = {"Low (<40°)": "steelblue",
               "Mid (40-55°)": "goldenrod",
               "High (55°+)": "tomato"}

for ax, (period_label, mask) in zip(axes, [
    ("Pre-storm (May 7-9)", pre_mask),
    ("Storm peak (May 11 00-12 UTC)", peak_mask)
]):
    subset = df_20m_raw[
        mask &
        (df_20m_raw["path_type"] == "day") &
        (df_20m_raw["dist_bucket"].isin(["1k-2k", "2k-4k"]))
    ]
    
    for mlat_label, color in mlat_colors.items():
        s = subset[subset["mlat_bin"] == mlat_label]["snr"]
        if len(s) < 100:
            continue
        ax.hist(s, bins=40, range=(-35, 5),
                color=color, alpha=0.5, density=True,
                label=f"{mlat_label} (n={len(s):,})")
    
    ax.set_xlabel("SNR (dB)")
    ax.set_title(f"20m Day DX — {period_label}")
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)

axes[0].set_ylabel("Density")
plt.suptitle("20m SNR Distribution by Magnetic Latitude\n"
             "Pre-storm vs Storm Peak, Gannon 2024",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
```
